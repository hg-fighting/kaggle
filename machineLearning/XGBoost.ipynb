```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Read the data
data = pd.read_csv('../input/melbourne-housing-snapshot/melb_data.csv')

# Select subset of predictors
cols_to_use = ['Rooms', 'Distance', 'Landsize', 'BuildingArea', 'YearBuilt']
X = data[cols_to_use]

# Select target
y = data.Price

# Separate data into training and validation sets
X_train, X_valid, y_train, y_valid = train_test_split(X, y)
```

```python
from xgboost import XGBRegressor

my_model = XGBRegressor()
my_model.fit(X_train, y_train)
```

```python
from sklearn.metrics import mean_absolute_error

predictions = my_model.predict(X_valid)
print("Mean Absolute Error: " + str(mean_absolute_error(predictions, y_valid)))
```

```python
my_model = XGBRegressor(n_estimators=500)
my_model.fit(X_train, y_train)
```

```python
my_model = XGBRegressor(
    n_estimators=500,
    early_stopping_rounds=5,  # 移到构造函数，消除警告
    eval_metric='mae',        # 显式指定评估指标，提升兼容性
    random_state=0,
    objective='reg:squarederror'  # 回归任务必加
)

# fit中仅保留eval_set和verbose，移除early_stopping_rounds
my_model.fit(
    X_train, y_train,
    eval_set=[(X_valid, y_valid)],
    verbose=False
)
```

```python
my_model = XGBRegressor(
    n_estimators=1000,
    learning_rate=0.05,
    early_stopping_rounds=5,  # 从fit中移到这里，消除警告
    eval_metric='mae',        # 显式指定评估指标，提升兼容性
    objective='reg:squarederror',  # 回归任务必加，避免默认值问题
    random_state=0            # 可选：固定随机种子，保证结果可复现
)

# fit方法中移除early_stopping_rounds，仅保留必要参数
my_model.fit(
    X_train, y_train,
    eval_set=[(X_valid, y_valid)],
    verbose=False
)
```

```python
my_model = XGBRegressor(
    n_estimators=1000,
    learning_rate=0.05,
    n_jobs=4,  # 多线程训练，保持不变
    early_stopping_rounds=5,  # 从fit移到这里，消除弃用警告
    eval_metric='mae',  # 显式指定评估指标，提升兼容性
    objective='reg:squarederror',  # 回归任务必加
    random_state=0  # 可选：固定随机种子，结果可复现
)

# fit方法中移除early_stopping_rounds，仅保留必要参数
my_model.fit(
    X_train, y_train,
    eval_set=[(X_valid, y_valid)],
    verbose=False
)
```
